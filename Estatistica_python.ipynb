```python

import numpy as np

#Conjunto de dados (amostra)
dados_idades = np.array([25, 30, 22, 35, 28])

#Calculo da amplitude total
amplitude_total = np.ptp(dados_idades)

print (f"O conjunto de dados é: {dados_idades}")
print (f"A amplitude total das idades é: {amplitude_total} anos")

# Calcula da Variancia. O parametro ddof indica
# Qual o tipo do calculo. ddof = 0 (polulaçã́o)
Variancia_amostral = np.var(dados_idades,
ddof = 1)

print(f"O conjunto de dados é: {dados_idades}")
print(f"A media das idades é: {np.mean(dados_idades)} anos")
print(f"A variancia amostral das idades é: {Variancia_amostral: .2f} anos²")
```

```python
# Desvio Padrão
import statistics as st

dados = [10, 20, 10, 30, 20, 10, 40]
desvio_padrao = st.stdev(dados)  # ≈ 11.83
print(desvio_padrao)
```

```python
import numpy as np

d = np.array([25, 30, 22, 35, 28])
print(f"Dados: {d}\nAmplitude: {np.ptp(d)} anos\nMédia: {np.mean(d):.1f} anos\nVariância: {np.var(d, ddof=1):.2f} anos²")
```
